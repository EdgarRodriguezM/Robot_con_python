```python
import random
def juego_adivinanza():
  numero_secreto = random.randint(1, 100)
  intentos = 0
  adivinado = False

  print("bienvenida al juego de adivinanza de números")
  print("Estoy pensando en unnumero entre el 1 y el 100")
  while  not adivinado:
    experimento = int(input("Adivina el numero: "))
    intentos +=1
    if experimento < numero_secreto:
      print("El número es mayor")
    elif experimento > numero_secreto:
      print("El numero es menor")
    else:
      adivinado = True
      print(f"Felicidades, adivinaste el nuemro en {intentos} intentos")
juego_adivinanza()
```

### Codigo del robot paseador


```python
import time
import random

class PaseadorMascotaRobot:
    def __init__(self, tiempo_objetivo_min=30):
        self.bateria_pct = 100
        self.tiempo_objetivo_min = tiempo_objetivo_min
        self.tiempo_transcurrido_min = 0
        self.correa_conectada = False
        self.perro_listo = False
        self.en_ruta = False
        self.ruta_establecida = []

    # -------------------------------------------------------------
    # 3.1 Preparar al perro para el paseo
    # -------------------------------------------------------------
    def preparar_perro(self):
        print("--- [3.1] Preparando al perro ---")
        print("• Colocando la correa al perro...")
        print("• Verificando ajuste de la correa... ¡Ajuste correcto!")
        self.perro_listo = True
        print("• Comprobando estado del perro... ¡El perro está listo!")
        print()

    # -------------------------------------------------------------
    # 3.2 Preparar el robot
    # -------------------------------------------------------------
    def preparar_robot(self):
        print("--- [3.2] Preparando el robot ---")
        print(f"• Verificando batería: {self.bateria_pct}% disponible.")
        if self.bateria_pct < 20:
            raise Exception("Batería insuficiente para iniciar el paseo.")

        print(f"• Tiempo de paseo programado: {self.tiempo_objetivo_min} minutos.")
        self.ruta_establecida = ["Frente de casa", "Esquina Norte", "Parque lateral", "Esquina Sur", "Regreso"]
        print(f"• Ruta trazada alrededor de la casa: {' -> '.join(self.ruta_establecida)}")

        if self.perro_listo:
            self.correa_conectada = True
            print("• Correa conectada exitosamente entre el robot y el perro.")
        else:
            raise Exception("No se puede conectar la correa: el perro no está listo.")
        print()

    # -------------------------------------------------------------
    # 3.4 Supervisar la seguridad (Evaluación en tiempo real)
    # -------------------------------------------------------------
    def evaluar_seguridad(self):
        if not self.correa_conectada:
            print("⚠️ ALERTA: ¡Correa desconectada! Deteniendo robot por seguridad.")
            return False

        # Simulación de eventos durante el monitoreo
        evento_aleatorio = random.choice(["OK", "OK", "OK", "obstaculo", "perro_detenido"])

        if evento_aleatorio == "obstaculo":
            print("  [Seguridad] Obstáculo o zona peligrosa detectada. Cambiando dirección/esquivando...")
            return True
        elif evento_aleatorio == "perro_detenido":
            print("  [Seguridad] El perro se ha detenido. Pausando marcha brevemente...")
            time.sleep(1)
            print("  [Seguridad] El perro reanuda la marcha. Continuando...")
            return True

        return True

    # -------------------------------------------------------------
    # 3.3 Realizar el recorrido
    # -------------------------------------------------------------
    def realizar_recorrido(self):
        print("--- [3.3 & 3.4] Iniciando paseo y supervisión continua ---")
        self.en_ruta = True
        paso_tiempo = 5  # Avance simula bloques de 5 minutos por iteración

        for punto in self.ruta_establecida:
            if not self.en_ruta:
                break

            print(f"📍 Avanzando hacia: {punto} (Velocidad adaptada al perro)")

            # Supervisar seguridad
            seguro = self.evaluar_seguridad()
            if not seguro:
                print("❌ Paseo interrumpido por fallo de seguridad.")
                return

            time.sleep(1)  # Simulación de tiempo de avance
            self.tiempo_transcurrido_min += paso_tiempo
            self.bateria_pct -= 2
            print(f"   --> Tiempo acumulado: {self.tiempo_transcurrido_min}/{self.tiempo_objetivo_min} min. Batería: {self.bateria_pct}%")

        print("• Recorrido completado con éxito.\n")

    # -------------------------------------------------------------
    # 3.5 Finalizar el paseo y regresar a casa
    # -------------------------------------------------------------
    def finalizar_paseo(self):
        print("--- [3.5] Finalizando paseo ---")
        if self.tiempo_transcurrido_min >= self.tiempo_objetivo_min:
            print(f"• Se han cumplido los {self.tiempo_objetivo_min} minutos de paseo.")

        print("• Llegada al punto de inicio (Casa).")
        self.en_ruta = False
        print("• Robot detenido por completo.")

        self.correa_conectada = False
        print("• Correa desconectada del perro de forma segura.")
        print("\n✅ Tarea completada: El perro está seguro en casa.")

    # -------------------------------------------------------------
    # Ejecución principal del proceso completo
    # -------------------------------------------------------------
    def ejecutar_paseo_completo(self):
        print("=== INICIANDO SISTEMA AUTÓNOMO DE PASEO DE MASCOTAS ===\n")
        self.preparar_perro()
        self.preparar_robot()
        self.realizar_recorrido()
        self.finalizar_paseo()

# Instanciar y correr en Colab
robot_paseador = PaseadorMascotaRobot(tiempo_objetivo_min=30)
robot_paseador.ejecutar_paseo_completo()
```

```python
import time

print("=== CONTROL DE PASEO AUTÓNOMO PARA PERROS ===")

# -------------------------------------------------------------
# 1. ENTRADA DE DATOS (INPUTS)
# -------------------------------------------------------------
# Pedimos el nivel de batería inicial
bateria = int(input("¿Qué porcentaje de batería tiene el robot? (0-100): "))

# Preguntamos por la preparación inicial
respuesta_correa = input("¿Le colocaste la correa al perro? (si/no): ").lower()
respuesta_perro = input("¿El perro está listo para salir? (si/no): ").lower()

# Validación de respuestas afirmativas
correa_conectada = (respuesta_correa == "si")
perro_listo = (respuesta_perro == "si")

# -------------------------------------------------------------
# 2. VERIFICACIÓN Y MANEJO DE ERRORES AL INICIAR
# -------------------------------------------------------------
print("\n--- Verificando estado inicial ---")

# Si la correa no está sujeta, el perro no está listo o la batería es menor a 30%
if not correa_conectada or not perro_listo or bateria < 30:
    print("❌ NO SE PUEDE INICIAR EL PASEO:")
    if not correa_conectada:
        print("   - Error: La correa no está colocada correctamente.")
    if not perro_listo:
        print("   - Error: El perro no está preparado.")
    if bateria < 30:
        print("   - Error: La batería es menor a 30% (" + str(bateria) + "%).")
    print("\nEl paseo ha sido cancelado por seguridad.")

else:
    print("✅ Todo correcto. Iniciando el paseo...")

    # Variables de control del paseo
    minuto = 0
    tiempo_maximo = 30
    perro_suelto = False

    # -------------------------------------------------------------
    # 3. BUCLE PRINCIPAL DEL PASEO (mientras minuto < 30)
    # -------------------------------------------------------------
    while minuto < tiempo_maximo:
        minuto = minuto + 5
        bateria = bateria - 5
        print("\n--------------------------------------------------")
        print("📍 Minuto " + str(minuto) + " de " + str(tiempo_maximo) + " | Batería: " + str(bateria) + "%")

        # Regla: Si la batería baja de 30% -> Regresar inmediatamente
        if bateria < 30:
            print("⚠️ Batería baja de 30%. Regresando inmediatamente a casa.")
            break

        # Regla: Si comienza a llover -> Regresar a casa
        lluvia = input("¿Comenzó a llover? (si/no): ").lower()
        if lluvia == "si":
            print("🌧️ Comenzó a llover. Cancelando paseo y regresando a casa.")
            break

        # Regla: Si se desvía del recorrido -> Retornar a la ruta
        desvio = input("¿El robot o perro se desviaron del recorrido? (si/no): ").lower()
        if desvio == "si":
            print("🔄 Recalculando... Retornando a la ruta programada.")

        # Regla: Si detecta vehículos aproximándose -> Detenerse
        vehiculo = input("¿Se aproxima un vehículo? (si/no): ").lower()
        if vehiculo == "si":
            print("🛑 Vehículo detectado. Deteniéndose hasta que pase...")
            time.sleep(1)

        # Regla: Si el perro hace sus necesidades -> Recogerlas
        necesidades = input("¿El perro hizo sus necesidades? (si/no): ").lower()
        if necesidades == "si":
            print("🧹 Recogiendo las necesidades del perro...")

        # Regla: Si encuentra un amigo -> Detenerse a dejarlo jugar
        amigo = input("¿El perro encontró a un amigo? (si/no): ").lower()
        if amigo == "si":
            print("🐕 Amigo detectado. Deteniéndose para dejarlo jugar un momento...")
            time.sleep(1)

        # Regla: Si la zona es segura -> Soltar al perrito para jugar
        zona_segura = input("¿La zona actual es segura (ej. parque cerrado)? (si/no): ").lower()
        if zona_segura == "si":
            perro_suelto = True
            print("🌳 Zona segura confirmada. Soltando al perro de la correa para jugar.")

            # Regla: Mientras el perrito traiga la pelota -> Lanzarla
            trae_pelota = input("¿El perro trae la pelota en el hocico? (si/no): ").lower()
            while trae_pelota == "si":
                print("🎾 Lanzando la pelota...")
                time.sleep(1)
                trae_pelota = input("¿El perro la volvió a traer? (si/no): ").lower()

            # Volver a colocar correa antes de continuar caminando
            print("🔗 Colocando la correa nuevamente para continuar el recorrido.")
            perro_suelto = False

        time.sleep(1) # Pausa simulada entre tramos del paseo

    # -------------------------------------------------------------
    # 4. FINALIZAR Y REGRESAR A CASA
    # -------------------------------------------------------------
    print("\n--------------------------------------------------")
    print("--- FIN DEL PASEO ---")
    print("• El robot y el perro han llegado de regreso a casa.")
    print("• Robot detenido.")

    if perro_suelto:
        print("• Se vuelve a enganchar al perro antes de entrar.")

    correa_conectada = False
    print("• Correa desconectada del perro.")
    print("✅ Proceso completado. Batería final: " + str(bateria) + "%")
```
